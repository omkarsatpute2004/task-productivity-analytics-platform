# 01. Data Cleaning Pipeline
This notebook demonstrates the data cleaning pipeline for the Task Management & Productivity Analytics Platform.
It loads raw task data from PostgreSQL database, performs deduplication, datetime parsing, missing value checks, and value range validations.

```python
import sys
import os
import pandas as pd
import numpy as np
from sqlalchemy import create_engine

# Add backend to path
sys.path.append(os.path.abspath('../backend'))
from app.analytics.queries import get_tasks_analytics_dataframe
from app.analytics.cleaning import clean_tasks_dataframe
from app.db.session import engine

print('Environment initialized successfully.')
```

## 1. Extract Raw Dataset from PostgreSQL Database

```python
from sqlalchemy.orm import sessionmaker
SessionLocal = sessionmaker(bind=engine)
db = SessionLocal()

raw_df = get_tasks_analytics_dataframe(db)
print(f'Raw tasks loaded: {len(raw_df)} rows, {len(raw_df.columns)} columns')
raw_df.head()
```

## 2. Execute Data Cleaning Pipeline

```python
clean_df, metadata = clean_tasks_dataframe(raw_df)
print('Data Cleaning Report:')
for key, val in metadata.items():
    print(f'  {key}: {val}')

print(f'\nCleaned DataFrame shape: {clean_df.shape}')
clean_df.info()
```

## 3. Verify Data Integrity & Column Data Types

```python
print('Missing Values Check:')
print(clean_df.isnull().sum())

print('\nSummary Statistics of Hours:')
print(clean_df[['estimated_hours', 'actual_hours']].describe())
```
