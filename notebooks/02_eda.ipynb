# 02. Exploratory Data Analysis (EDA)
This notebook conducts Exploratory Data Analysis on the task productivity dataset.
It examines distributions by status, priority, category, user department, and completion trends over time.

```python
import sys
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Add backend to path
sys.path.append(os.path.abspath('../backend'))
from app.analytics.queries import get_tasks_analytics_dataframe
from app.analytics.cleaning import clean_tasks_dataframe
from app.analytics.features import enrich_task_features
from app.analytics.calculations import (
    compute_summary_kpis,
    compute_status_distribution,
    compute_priority_analysis,
    compute_category_analysis,
    compute_user_productivity,
    compute_department_analysis,
    compute_completion_trend
)
from app.db.session import engine
from sqlalchemy.orm import sessionmaker

SessionLocal = sessionmaker(bind=engine)
db = SessionLocal()
df_raw = get_tasks_analytics_dataframe(db)
df_clean, _ = clean_tasks_dataframe(df_raw)
df = enrich_task_features(df_clean)
print(f'Enriched dataset ready with {len(df)} records.')
```

## 1. High-Level Summary KPIs

```python
kpis = compute_summary_kpis(df)
for k, v in kpis.items():
    print(f'{k}: {v}')
```

## 2. Status Distribution & Priority Breakdown

```python
status_dist = compute_status_distribution(df)
print('Status Distribution:')
print(pd.DataFrame(status_dist))

prio_dist = compute_priority_analysis(df)
print('\nPriority Breakdown:')
print(pd.DataFrame(prio_dist))
```

## 3. Category & Department Analysis

```python
cat_analysis = compute_category_analysis(df)
print('Top Categories by Task Volume:')
print(pd.DataFrame(cat_analysis))

dept_analysis = compute_department_analysis(df)
print('\nDepartmental Breakdown:')
print(pd.DataFrame(dept_analysis))
```

## 4. Completion Trend (Monthly)

```python
trend = compute_completion_trend(df, interval='monthly')
trend_df = pd.DataFrame(trend)
print(trend_df)
```
