# 04. Machine Learning Data Preparation & Leakage Prevention
This notebook inspects the task productivity dataset for supervised machine learning.
It establishes target definitions (`is_late` classification target, `completion_days` regression target), prevents data leakage by excluding post-completion attributes (`completed_at`, `actual_hours`, `status`), checks class distributions, and configures train/test splitting.

```python
import sys
import os
import pandas as pd
import numpy as np
from sqlalchemy import create_engine

sys.path.append(os.path.abspath('../backend'))
from app.analytics.queries import get_tasks_analytics_dataframe
from app.ml.features import extract_ml_dataset, FEATURE_COLUMNS
from app.db.session import engine
from sqlalchemy.orm import sessionmaker

SessionLocal = sessionmaker(bind=engine)
db = SessionLocal()
raw_df = get_tasks_analytics_dataframe(db)
print(f'Raw dataset size: {len(raw_df)} rows')
```

## 1. Extract Completed Historical Training Set & Prevent Data Leakage

```python
df, stats = extract_ml_dataset(raw_df)
print('Dataset Extraction Summary:')
for k, v in stats.items():
    print(f'  {k}: {v}')

print(f'\nSelected Prediction Features ({len(FEATURE_COLUMNS)}): {FEATURE_COLUMNS}')
```

## 2. Inspect Target Class Balance & Feature Distributions

```python
print('Overdue Target (is_late) Class Balance:')
print(df['is_late'].value_counts(normalize=True).rename(index={0: 'On-Time (0)', 1: 'Late (1)'}) * 100)

print('\nCompletion Days Target Summary:')
print(df['completion_days'].describe())
```
