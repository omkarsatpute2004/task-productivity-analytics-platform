# 06. Completion Time Prediction (Regression)
This notebook evaluates regression models (Linear Regression vs Random Forest Regressor) to predict task completion duration in days.
It measures MAE (Mean Absolute Error in days), RMSE, and R^2 scores.

```python
import sys
import os
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split

sys.path.append(os.path.abspath('../backend'))
from app.analytics.queries import get_tasks_analytics_dataframe
from app.ml.features import extract_ml_dataset, FEATURE_COLUMNS
from app.ml.regression import train_and_evaluate_regressors
from app.ml.evaluation import extract_feature_importance
from app.db.session import engine
from sqlalchemy.orm import sessionmaker

SessionLocal = sessionmaker(bind=engine)
db = SessionLocal()
raw_df = get_tasks_analytics_dataframe(db)
df, _ = extract_ml_dataset(raw_df)
print(f'Training dataset ready with {len(df)} records.')
```

## 1. Train and Evaluate Regression Models

```python
df_reg = df.dropna(subset=['completion_days'])
X = df_reg[FEATURE_COLUMNS]
y = df_reg['completion_days']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)

best_pipeline, metrics, comparison = train_and_evaluate_regressors(X_train, y_train, X_test, y_test)
print(f'Selected Model: {comparison["selected_model"]}')
print('Metrics Comparison:')
print(pd.DataFrame(comparison['candidate_metrics']))
```
